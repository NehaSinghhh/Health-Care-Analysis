```python


# Do males and females have different visit patterns?


gender_visits = df.groupby("gender")["visits"].mean()

print("\nAverage Visits by Gender:")
print(gender_visits)

plt.figure(figsize=(6, 5))

sns.barplot(
    data=df,
    x="gender",
    y="visits"
)

plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Visits")
plt.show()
```

```python

# Does income influence healthcare utilization?


income_visits = df.groupby("income")["visits"].mean()

plt.figure(figsize=(8, 5))

plt.scatter(
    income_visits.index,
    income_visits.values
)

plt.title("Income vs Average Doctor Visits")
plt.xlabel("Income")
plt.ylabel("Average Doctor Visits")
plt.grid()
plt.show()


# Correlation
print("\nCorrelation between Income and Visits:")
print(df["income"].corr(df["visits"]))

```

```python


# Does number of illnesses affect doctor visits?


illness_visits = df.groupby("illness")["visits"].mean()

print("\nAverage Visits by Number of Illnesses:")
print(illness_visits)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="illness",
    y="visits"
)

plt.title("Illness vs Doctor Visits")
plt.xlabel("Number of Illnesses")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python

# Do chronic conditions increase doctor visits?


nchronic_visits = df.groupby("nchronic")["visits"].mean()

print("\nVisits According to Recent Chronic Condition:")
print(nchronic_visits)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="nchronic",
    y="visits"
)

plt.title("Chronic Condition vs Doctor Visits")
plt.xlabel("Chronic Condition")
plt.ylabel("Average Doctor Visits")
plt.show()


# Long-term chronic condition
lchronic_visits = df.groupby("lchronic")["visits"].mean()

print("\nVisits According to Long-Term Chronic Condition:")
print(lchronic_visits)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="lchronic",
    y="visits"
)

plt.title("Long-Term Chronic Condition vs Doctor Visits")
plt.xlabel("Long-Term Chronic Condition")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python


# Is reduced activity associated with more visits?


reduced_visits = df.groupby("reduced")["visits"].mean()

print("\nAverage Visits by Reduced Activity:")
print(reduced_visits)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="reduced",
    y="visits"
)

plt.title("Reduced Activity vs Doctor Visits")
plt.xlabel("Days of Reduced Activity")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python

# Does private healthcare coverage affect visits?


private_visits = df.groupby("private")["visits"].mean()

print("\nAverage Visits by Private Healthcare Coverage:")
print(private_visits)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="private",
    y="visits"
)

plt.title("Private Healthcare Coverage vs Doctor Visits")
plt.xlabel("Private Healthcare")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python

# Does free healthcare coverage affect visits?


freepoor_visits = df.groupby("freepoor")["visits"].mean()

print("\nAverage Visits - Free Healthcare for Low Income:")
print(freepoor_visits)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="freepoor",
    y="visits"
)

plt.title("Free Healthcare for Low Income vs Doctor Visits")
plt.xlabel("Free Healthcare")
plt.ylabel("Average Doctor Visits")
plt.show()
```

```python

# Which numerical factors have the strongest
# relationship with doctor visits?


numeric_columns = [
    "visits",
    "age",
    "income",
    "illness",
    "reduced",
    "health"
]

correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Healthcare Variables")
plt.show()
```
